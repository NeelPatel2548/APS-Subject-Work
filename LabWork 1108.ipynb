```python
# Complete the exercise and write the learning in one of the cell
# 1. Problem1: A fair coin is tossed 100 times.Calculate:
# Number of heads
# Number of tails
# Experimental probability of Head
# Experimental probability of Tail

import random as rm
lis = []
for i in range(0,100):
    outcome = rm.randrange(0, 2) 
    lis.append(outcome)
    # print(outcome)

# print(len(lis))
# HEAD = 0 AND TAIL = 1
head_cout = 0  
tail_count = 0

for val in lis:
    if val == 0:
        # print("Head")
        head_cout += 1
    if val == 1:
        # print("Tail")
        tail_count += 1

print("No of Heads: ",head_cout)        #number of heads
print("No of Tails: ",tail_count)       #number of tails

#finding prob.          prob. = total outcome / total trial

prob_head = (head_cout / 100)
print(prob_head)


prob_tail = (tail_count / 100)
print(prob_tail)
```

```python
# 2. Problem 2: check Effect of the number of trials
# Repeat the coin-toss experiment for:  

import random as rm
class coins:

    def count_coin(self, trial):
        lis = []

        for i in range(0,trial):
            outcome = rm.randrange(0, 2) 
            lis.append(outcome)

        head_cout = 0  
        tail_count = 0

        for val in lis:
            if val == 0:
                # print("Head")
                head_cout += 1
            elif val == 1:
                # print("Tail")
                tail_count += 1

        # print("No of Heads: ",head_cout)        #number of heads
        # print("No of Tails: ",tail_count)       #number of tails

        #finding prob.          prob. = total outcome / total trial

        prob_head = (head_cout / trial)
        print("Probability of getting HEAD : ", prob_head)


        prob_tail = (tail_count / trial)
        print("Probability of getting TAIL : ",prob_tail)


cn = coins()

cn.count_coin(10)
cn.count_coin(100)
cn.count_coin(1000)
cn.count_coin(10000)


#Conclusion wriiten by gpt:

# From the results of tossing a fair coin different numbers of times, we can observe that the probability of getting Heads and Tails becomes closer to 0.5 as the number of tosses increases.

# 10 tosses: Heads = 0.80, Tails = 0.20 → large variation
# 100 tosses: Heads = 0.53, Tails = 0.47 → closer to 0.5
# 1,000 tosses: Heads = 0.514, Tails = 0.486 → very close to 0.5
# 10,000 tosses: Heads = 0.5047, Tails = 0.4953 → almost exactly 0.5

# Therefore, as the number of trials increases, the experimental probability approaches the theoretical probability.
```

```python
# 3. Rolling a die: A fair die is rolled 6,000 times.
# Calculate the experimental probability of each outcome: S={1,2,3,4,5,6}

import random as rm
lis = []
for i in range(0,6000):
    outcome = rm.randrange(1, 7) 
    lis.append(outcome)

# print(lis)
# print(len(lis))

count_1 = 0
count_2 = 0
count_3 = 0
count_4 = 0
count_5 = 0
count_6 = 0


for val in lis:
    if val == 1:
        count_1 += 1
    if val == 2:
        count_2 += 1
    if val == 3:
        count_3 += 1
    if val == 4:
        count_4 += 1
    if val == 5:
        count_5 += 1
    if val == 6:
        count_6 += 1

# print("No of 1: ",count_1)  # no of time 1 appears
# print("No of 2: ",count_2)  # no of time 2 appears
# print("No of 3: ",count_3)  # no of time 3 appears
# print("No of 4: ",count_4)  # no of time 4 appears
# print("No of 5: ",count_5)  # no of time 5 appears
# print("No of 6: ",count_6)  # no of time 6 appears 

#finding prob.          prob. = total outcome / total trial

prob_count1 = (count_1 / 6000)
print("Probability of 1: ",round(prob_count1,4))

prob_count2 = (count_2 / 6000)
print("Probability of 2: ",round(prob_count2,4))

prob_count3 = (count_3 / 6000)
print("Probability of 3: ",round(prob_count3,4))

prob_count4 = (count_4 / 6000)
print("Probability of 4: ",round(prob_count4,4))

prob_count5 = (count_5 / 6000)
print("Probability of 5: ",round(prob_count5,4))

prob_count6 = (count_6 / 6000)
print("Probability of 6: ",round(prob_count6,4))
```

```python
# 4. Addition rule for disjoint events
# A die is rolled 10,000 times.
# Define: A=getting 1, B= getting 6, check the events are joint or disjoint
# Verify: P(A∪B)=P(A)+P(B)

prob_count1 = (count_1 / 6000)
print("Probability of A: ",round(prob_count1,4))

prob_count6 = (count_6 / 6000)
print("Probability of B: ",round(prob_count6,4))

print("P(A∪B)=P(A)+P(B): ", prob_count1 + prob_count6 )


# Conclusion: these events are disjoint because both event can not happen at the same time. either we got 1 or 6 not both at same time.
```

```python
# 5. Addition rule for overlapping events
# For one die roll, define: A=getting an even number A={2,4,6} and B=getting a number greater than 3 B={4,5,6}
# Verify: P(A∪B)=P(A)+P(B)−P(A∩B)
import random as rm
lis = []
for i in range(0,6000):
    outcome = rm.randrange(1, 7) 
    lis.append(outcome)


count_A = 0
count_B = 0
count_Join = 0
count_4A = 0
count_6A = 0


for val in lis:
    if val == 2 or val == 4 or val == 6:
        count_A += 1
    if val == 4 or val == 5 or val == 6:
        count_B += 1
    if val == 4 or val == 6:
        count_Join += 1


P_A = count_A / 6000
P_B = count_B / 6000
P_inter = count_Join / 6000

# A ∪ B = {2,4,5,6}
P_union = P_A + P_B - P_inter

print("P(A) =", P_A)
print("P(B) =", P_B)
print("P(A ∩ B) =", P_inter)
print("P(A ∪ B) =", P_union)
```

```python
# 6. Multiplication rule for independent events
# A coin is tossed and a die is rolled 10,000 times.
# Define:
#  obtaining Head
#  obtaining an even number
# Verify: P(A∩B)=P(A)P(B)

class prob:

    def count_coin(self, coin):
        list_coint = []

        for i in range(0,coin):
            out = rm.randrange(0,2)
            list_coint.append(out)

        head = 0
        for co in list_coint:
            if co == 0: #0 = head
                head += 1 

        return head        #It return no of time head comes


    def count_die(self, die):

        list_die = []

        for i in range(0,die):
            die_out = rm.randrange(1,7)
            list_die.append(die_out)

        die_even = 0
        for valu in list_die:
            if valu == 2 or valu == 4 or valu == 6: 
                die_even += 1

        return die_even              #It return no of time even numbers come



pr = prob()

trial = 10000
get_head = pr.count_coin(trial)
# print("No. of time head appears: ", get_head)

get_die = pr.count_die(trial)
# print("No. of time even numbers appears: ", get_die)

print("Prob. of Event A : ", get_head / trial)
print("Prob. of Event B : ", get_die / trial)

print("P(A∩B)=P(A)P(B): ", (get_head / trial) * (get_die / trial)  )

```
