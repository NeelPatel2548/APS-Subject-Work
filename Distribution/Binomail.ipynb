what it do: This line generates random samples from a Binomial distribution using NumPy. Since n=1, it behaves like a Bernoulli trial.

```python
# Import Binomial distribution from scipy
from scipy.stats import binom

# Total number of trials
n = 10

# Probability of success in each trial
p = 0.30

# Create a Binomial distribution object
X = binom(n, p)

# Possible number of successes: 0 to n
support = list(range(n + 1))
print("Values of X:", support)

# Display the distribution name and parameters
print("Distribution: Binomial(n =", n, ", p =", p, ")")

# Probability of exactly 0 successes
print("P(X = 0):", X.pmf(0))

# Probability of exactly 3 successes
print("P(X = 3):", X.pmf(3))

# Probability of exactly 10 successes
print("P(X = 10):", X.pmf(10))

# Sum of probabilities for all possible values
print("PMF sum:", sum(X.pmf(x) for x in support))
```

```python
# Generate 10 experiments, each with 10 trials
samples = np.random.binomial(n=n, p=p, size=10)

# Calculate theoretical and empirical PMF
for x in range(n + 1):

    # Theoretical probability using Binomial distribution
    theoretical = X.pmf(x)

    # Empirical probability from simulated observations
    empirical = np.mean(samples == x)

    print("\nX =", x)
    print("Theoretical PMF:", round(theoretical,4))
    print("Empirical PMF:", empirical)
```
