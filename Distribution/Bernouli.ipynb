<h3>samples = np.random.binomial(n=1, p=0.30, size=100)</h3>

if we change n value to more than 1 than it generate for binomial

what it do: This line generates random samples from a Binomial distribution using NumPy. Since n=1, it behaves like a Bernoulli trial.

```python
# Import Bernoulli distribution from scipy
from scipy.stats import bernoulli

# Probability of success
p = 0.30

# Create a Bernoulli distribution object
X = bernoulli(p)

# Possible values: 0 = Failure, 1 = Success
support = [0, 1]
print("Values of X:", support)

# Display the distribution name and parameter
print("Distribution: Bernoulli(p =", p, ")")

# Calculate probability of failure
print("P(X = 0):", round(X.pmf(0),3))

# Calculate probability of success
print("P(X = 1):", X.pmf(1))

# Sum of all probabilities must equal 1
print("PMF:", X.pmf(0) + X.pmf(1))
```

```python
import numpy as np
from scipy.stats import bernoulli

# Define the theoretical probability of success
p = 0.5

# Create the theoretical Bernoulli distribution
X = bernoulli(p)

# Generate 10 random observations
samples = np.random.binomial(n=1, p=p, size=10)

print("Samples:", samples)

# Calculate theoretical and empirical PMF
for x in [0, 1]:

    # Probability from the mathematical distribution
    theoretical = X.pmf(x)

    # Probability from the simulated observations
    empirical = np.mean(samples == x)

    print("\nX =", x)
    print("Theoretical PMF:", theoretical)
    print("Empirical PMF:", empirical)
```

```python
import numpy as np

# 1. Standard Normal Distribution
X1 = np.random.randn(1, 5)
print("1. Normal Distribution:")
print(X1)

# 2. Uniform Distribution: numbers between 0 and 1
X2 = np.random.rand(1, 5)
print("\n2. Uniform Distribution (0 to 1):")
print(X2)

# 3. Random Integers: dice rolls from 1 to 6
X3 = np.random.randint(1, 7, size=5)
print("\n3. Dice Rolls:")
print(X3)

# 4. Bernoulli Distribution: one trial, success or failure
X4 = np.random.binomial(n=1, p=0.3, size=5)
print("\n4. Bernoulli Distribution:")
print(X4)

# 5. Binomial Distribution: 10 trials per experiment
X5 = np.random.binomial(n=10, p=0.3, size=5)
print("\n5. Binomial Distribution:")
print(X5)

# 6. Hypergeometric Distribution: sampling without replacement
X6 = np.random.hypergeometric(
    ngood=3, nbad=7, nsample=4, size=5
)
print("\n6. Hypergeometric Distribution:")
print(X6)

# 7. Normal Distribution: mean 50, standard deviation 10
X7 = np.random.normal(loc=50, scale=10, size=5)
print("\n7. Normal Distribution (mean=50, SD=10):")
print(X7)
```
