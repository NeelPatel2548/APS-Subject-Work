A hypergeometric distribution counts successes when selecting items without replacement from a finite population.

```python
import random

balls = ["R"] * 3 + ["B"] * 7
random.shuffle(balls)

sample = random.sample(balls, 4)
success = sample.count("R")

print("Selected balls:", sample)
print("Number of red balls:", success)
```

```python
# Import Hypergeometric distribution from scipy
from scipy.stats import hypergeom

# Total number of items in the population
N = 10

# Number of success items (defective items)
K = 3

# Number of items selected without replacement
n = 4

# Create a Hypergeometric distribution object
X = hypergeom(N, K, n)

# Possible number of defective items selected
# Minimum is 0; maximum is min(K, n)
support = list(range(0, min(K, n) + 1))
print("Values of X:", support)

# Display the distribution name
print("Distribution: Hypergeometric")

# Probability of selecting exactly 0 defective items
print("P(X = 0):", X.pmf(0))

# Probability of selecting exactly 1 defective item
print("P(X = 1):", X.pmf(1))

# Probability of selecting exactly 2 defective items
print("P(X = 2):", X.pmf(2))

# Sum of probabilities for all possible values
print("PMF sum:", sum(X.pmf(x) for x in support))
```
