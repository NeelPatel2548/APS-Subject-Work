```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
class bday:

    def count_prob(self, ppl):
        prob = 1
        list_x = []
        list_y = []

        for i in range(1,ppl + 1):
            # print("for ", ppl, "Bday probability is :")

            prob *= (365 - i + 1) / 365
            same_prob = 1 - prob
            list_x.append(i)
            list_y.append(same_prob)
            print("Probability of same birthday for ", i , "no. of ppl is: ",round(same_prob, 5))

        xpoints = np.array(list_x)
        ypoints = np.array(list_y)

        plt.plot(xpoints, ypoints)
        plt.show()


bd = bday()
inp = 100
# inp = input('Enter for howmany people u want to calculate: ')
bd.count_prob(int(inp))
```
