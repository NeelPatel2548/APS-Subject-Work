```python
import numpy as np
from scipy.stats import (
    norm, ttest_1samp, ttest_ind, ttest_rel,
    chi2_contingency )

alpha = 0.05

def decision(p_value):
    print("P-value:", p_value)

    if p_value <= alpha:
        print("Reject H0")
    else:
        print("Fail to reject H0")
```

### One-sample Z-test

```python
data = [52, 55, 48, 51, 54]

mu = 50       # Population mean
sigma = 5     # Population standard deviation
n = len(data)

x_bar = np.mean(data)

z = (x_bar - mu) / (sigma / np.sqrt(n))

# Two-tailed p-value
p_value = 2 * norm.sf(abs(z))

print("Z-statistic:", z)
decision(p_value)
```

### One-sample T-test

```python
data = [45, 48, 52, 50, 55, 49, 53, 51]

t_stat, p_value = ttest_1samp(data, popmean=50)

print("T-statistic:", t_stat)
decision(p_value)
```

### Independent two-sample T-test


Use: Compare the means of two independent groups.

```python
class1 = [45, 50, 55, 60, 52]
class2 = [60, 65, 70, 68, 72]

t_stat, p_value = ttest_ind(
    class1, class2,
    equal_var=False
)

print("T-statistic:", t_stat)
decision(p_value)
```

### Paired T-test



Use: Compare measurements from the same people or objects before and after a treatment.

```python
before = [50, 55, 60, 65, 70]
after  = [55, 58, 65, 68, 75]

t_stat, p_value = ttest_rel(before, after)

print("T-statistic:", t_stat)
decision(p_value)
```

### Chi-square test of independence



Use: Check whether two categorical variables are associated.

```python
observed = [
    [20, 30],
    [40, 10]
]

chi_stat, p_value, dof, expected = chi2_contingency(observed)

print("Chi-square:", chi_stat)
print("Degrees of freedom:", dof)
print("Expected frequencies:\n", expected)

decision(p_value)
```
